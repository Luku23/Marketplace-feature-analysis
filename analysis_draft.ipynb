```python
# Import necessary libraries
import numpy as np
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read the csv file and convert to dataframe using pandas
df = pd.read_csv("Pricing Push Ups - data.csv")

# Check data
df.head()
```

```python
clean_df = df.fillna({"revenue_from_push_ups": 0})

clean_df.sort_values('revenue_from_push_ups')
```

```python
NUM_OF_LISTINGS = clean_df['number_of_listings'].to_numpy()

plt.xscale('log')
plt.title('number_of_listings (log scale)')
plt.boxplot(NUM_OF_LISTINGS, vert=False)
```

```python
XY = list(zip(clean_df['number_of_listings'].to_numpy(), clean_df['revenue_from_push_ups'].to_numpy()))

print(XY)
```

```python
x, y = zip(*XY)

plt.scatter(x, y)
plt.ylabel('revenue from push ups')
plt.xlabel('number of listings')
plt.xscale('log')
plt.yscale('log')
plt.savefig('listings-revenue.png')
```

```python
import seaborn as sns
clean_df2 = clean_df[clean_df.columns[2:5]]
corr = clean_df2.corr()
plt.figure()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
plt.savefig('corr.jpg')

print(corr['number_of_listings'].sort_values(ascending=False))
```

```python
revenue_from_push_ups = np.int64(clean_df2['revenue_from_push_ups'].to_numpy())
number_of_push_ups = np.int64(revenue_from_push_ups / 2)
number_of_push_ups
```

```python
number_of_listings = np.int64(clean_df2['number_of_listings'].to_numpy())
number_of_listings
```

```python
push_up_rate = number_of_push_ups/number_of_listings*100
```

```python
clean_df2['push_up_rate'] = push_up_rate
```

```python
clean_df2.sort_values('push_up_rate', ascending=False)
```

```python
number_of_push_ups.sum() / number_of_listings.sum() * 100
```

```python
(clean_df2['revenue_from_push_ups'].sum() / 2) / clean_df2['number_of_listings'].sum() * 100
```

```python
X_NUMBER_OF_LISTINGS = clean_df2['number_of_listings']
Y_PUSH_UP_RATE = clean_df2['push_up_rate']
BUBBLE_SIZE = clean_df2['revenue_from_push_ups'] / 30

plt.figure(figsize=(7, 6))
plt.margins(x=0.07)

plt.scatter(
    X_NUMBER_OF_LISTINGS,
    Y_PUSH_UP_RATE,
    s=BUBBLE_SIZE,
    alpha=0.4
    )
plt.xlabel('Number of listings')
plt.ylabel('Push-up rate (%)')
plt.title('Push-up rate vs. number of listings by subcategory')
plt.xscale('log')
plt.yscale('log')

# Bubble-size legend
revenue_values = [100, 1000, 10000, 50000]

for revenue in revenue_values:
    plt.scatter(
        [],
        [],
        s=revenue/30,
        alpha=0.5,
        label=f'€{revenue:,}'
    )

plt.legend(title='Push-up revenue')

plt.savefig('push_up_rate_vs_listings.jpg', dpi=300, bbox_inches='tight')
plt.show()
```
